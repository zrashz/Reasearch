2. Upload/install update and test


```python
from google.colab import files
from pathlib import Path
import subprocess, sys, tempfile, zipfile
PROJECT = Path('/content/drive/MyDrive/scientific-news-verification')
uploaded = files.upload()
matches = []
for name in uploaded:
    if zipfile.is_zipfile(name):
        with zipfile.ZipFile(name) as archive:
            if 'install_gpu_update.py' in archive.namelist():
                matches.append(name)
assert len(matches) == 1, 'Upload Scientific_News_GPU_Development_Update.zip.'
bundle = Path(tempfile.mkdtemp(prefix='gpu_update_'))
with zipfile.ZipFile(matches[0]) as archive:
    for item in archive.infolist():
        assert (bundle/item.filename).resolve().is_relative_to(bundle.resolve()), 'Unsafe ZIP path.'
    archive.extractall(bundle)
for command in [
    [sys.executable, str(bundle/'install_gpu_update.py'), '--project', str(PROJECT)],
    [sys.executable, '-m', 'unittest', 'discover', '-s', 'tests', '-v'],
]:
    result = subprocess.run(command, cwd=PROJECT, capture_output=True, text=True)
    print(result.stdout)
    print(result.stderr)
    result.check_returncode()
```

3. Run the first GPU paper


```python
from pathlib import Path
from datetime import datetime, timezone
import subprocess, sys
PROJECT = Path('/content/drive/MyDrive/scientific-news-verification')
BATCH = PROJECT/'private/eligibility_screening/batch_20261119T080435821348Z'
SCREEN = BATCH/'public_fulltext_screen_20261119T081014931038Z'
CASE = SCREEN/'train14460_condition_inputs_20261119T091631988786Z'
stamp = datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S%fZ')
OUT = PROJECT/f'private/gpu_development_runs/train14460_E2_{stamp}'
command = [sys.executable, '-u', str(PROJECT/'run_gpu_development_v1.py'), '--project', str(PROJECT), '--case', str(CASE), '--source-xml', str(SCREEN/'xml/train_14460.xml'), '--metadata', str(BATCH/'screening_metadata.csv'), '--output', str(OUT), '--conditions', 'E2']
print('Starting one GPU development paper, E2, with V0–V3.', flush=True)
print('Saved results:', OUT, flush=True)
with subprocess.Popen(command, cwd=PROJECT, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1) as process:
    for line in process.stdout:
        print(line, end='', flush=True)
    code = process.wait()
print('Exit code:', code)
print('Saved:', OUT)
if code:
    print('Stopped with saved events. Share the output; do not rerun this cell.')
```
